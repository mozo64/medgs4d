# MedGS4D results browser

This notebook is a thin client for one completed run. It does not prepare DICOM data, train models, or redefine the renderer.


```python
from pathlib import Path

from medgs4d.results import load_run, print_run_summary
from medgs4d.visualization import (
    show_breathing_cycle_browser,
    show_slice_browser,
    show_training_curves,
)
```

## Select a run

Set either `RUN_DIR` directly or resolve it from `RESULTS_ROOT`, `STUDY_NAME`, and `RUN_NAME`.


```python
# Option A: direct run path.
RUN_DIR = Path("/absolute/path/to/results/<study-name>/<run-name>")

# Option B: set RUN_DIR = None and resolve by names.
RESULTS_ROOT = Path("/absolute/path/to/results")
STUDY_NAME = "<study-name>"
RUN_NAME = "<run-name>"
DEVICE = "cuda"

run = (
    load_run(run_dir=RUN_DIR)
    if RUN_DIR is not None
    else load_run(
        results_root=RESULTS_ROOT,
        study_name=STUDY_NAME,
        run_name=RUN_NAME,
    )
)
print_run_summary(run)
```

## Saved metrics


```python
display(run.training_history.tail())

if run.validation_history is not None:
    display(run.validation_history)

if run.per_phase_metrics is not None:
    display(run.per_phase_metrics)

run.overall_metrics
```

## Training and validation curves


```python
show_training_curves(run)
```

## Interactive phase and slice browser


```python
show_slice_browser(run, device=DEVICE)
```

## Interactive full respiratory cycle


```python
show_breathing_cycle_browser(run, device=DEVICE)
```
